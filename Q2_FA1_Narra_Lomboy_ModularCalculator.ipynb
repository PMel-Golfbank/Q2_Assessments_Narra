```python
# This is a simple modular mini-calculator program that serves as the FA output.
# It asks the user to provide the numbers to work on and what operation to perform.

# Calculates the sum of two quantities.
def add_numbers(num1, num2):
  return num1 + num2

# Calculates the difference of two quantities.
def subtract_numbers(num1, num2):
  return num1 - num2

# Calculates the product of two quantities.
def multiply_numbers(num1, num2):
  return num1 * num2

# Calculates the quotient of two quantities. This contains a safeguard against division by zero.
def divide_numbers(num1, num2):
  if num2 == 0:
    return "Cannot divide by zero."
  return num1 / num2

# Prompt the user for the first and second numbers.
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

print("\nChoose operation:\n1 - Addition\n2 - Subtraction\n3 - Multiplication\n4 - Division")

# Prompt the user for their choice.
# Then, perform the corresponding operation through an if-elif selection.
choice = int(input("\nEnter choice: "))
result = None

if choice == 1:
  result = add_numbers(num1, num2)
elif choice == 2:
  result = subtract_numbers(num1, num2)
elif choice == 3:
  result = multiply_numbers(num1, num2)
elif choice == 4:
  result = divide_numbers(num1, num2)

print("\nResult:", result)
```
